```python
import pandas as pd
```

```python
df_exams = pd.read_csv('studentperformance.csv')
```

```python
df_exams.head()
```

index

```python
import numpy as np
```

```python
import random
```

```python
#creating non repetetive index
new_index = np.arange(0,20)
```

```python
#shuffling indexes
random.shuffle(new_index)
```

```python
#creating new column with new index
df_exams['new_index'] = new_index
```

```python
df_exams
```

set index

```python
df_exams.set_index('new_index',inplace = True)
```

```python
df_exams.head()
```

sort index

```python
#sort dataframe by new index
df_exams.sort_index()
```

```python
#sorting in descending 
df_exams.sort_index(ascending = False)
```
