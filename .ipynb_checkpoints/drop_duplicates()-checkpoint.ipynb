```python
import pandas as pd
```

```python
df_laptops = pd.read_csv('laptop_price.csv',encoding = 'unicode_escape')
```

```python
df_laptops.head()
```

```python
df_laptops.drop_duplicates(['Company'])[['Company','Price_euros']].value_counts('Company')
```

```python
df_laptops = df_laptops.sort_values(['Company','Price_euros'])
```

```python
df_laptops.drop_duplicates(['Company'],keep='first')[['Company','Price_euros']] #cheap
```

```python
df_laptops.drop_duplicates(['Company'],keep='last',ignore_index=True)[['Company','Price_euros']] #expensive , inplace = true 
```
