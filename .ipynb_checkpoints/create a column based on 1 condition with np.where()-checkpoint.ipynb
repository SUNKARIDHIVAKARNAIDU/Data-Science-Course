```python
import pandas as pd
```

```python
df_laptops = pd.read_csv('laptop_price.csv', encoding = 'unicode_escape')
```

```python
df_laptops.head()
```

condition 1  using np.where()

```python
import numpy as np
```

```python
#array based on price tiers
np.where(df_laptops['Price_euros']>2000,'Expensive','cheap')
```

```python
#set into a new column
df_laptops['Price_tiers'] = np.where(df_laptops['Price_euros']>2000,'Expensive','cheap')
```

```python
#show first five columns
df_laptops.head()
```

```python
df_laptops.value_counts('Price_tiers')
```
