merge

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'id':['A','B','C','D',],
    'age':['30','23','25','22']})
df2 = pd.DataFrame({
    'id': ['C', 'D', 'E', 'F'],
    'job': ['Doctor', 'Statistician','Accountant', 'Developer']})
```

```python
df1
```

```python
df2
```

```python
df_inner_join = df1.merge(df2,on = 'id',how = 'inner')
```

```python
df1.shape
```

```python
df2.shape
```

```python
df_inner_join.shape
```

exercise

```python
df_movies = pd.read_csv('IMDb movies.csv', low_memory=False)
df_ratings = pd.read_csv('IMDb ratings.csv')

# select columns
df_movies = df_movies[['imdb_title_id', 'title', 'year',
                       'genre', 'country']]

df_ratings = df_ratings[['imdb_title_id', 'total_votes', 'mean_vote']]
```

```python
df_movies
```

```python
df_ratings
```

```python
df_movies.merge(df_ratings, on = 'imdb_title_id', how = 'inner')
```

```python
pd.merge(df_movies,df_ratings, on = 'imdb_title_id' )
```

full join 

```python
df1.merge(df2,on = 'id',how = 'outer')
```

```python
df_movies.merge(df_ratings, on = 'imdb_title_id', how = 'outer')
```

```python
pd.merge(df_movies,df_ratings, on = 'imdb_title_id' )
```

exclusive full join

```python
df1.merge(df2, on = 'id',how = 'outer')
```

```python
df1.merge(df2, on = 'id',how = 'outer',indicator = True)
```

```python
df1.merge(df2, on = 'id',how = 'outer',indicator = True).query("_merge == 'left_only' or _merge == 'right_only' ")
```

exercise

```python
df_movies.merge(df_ratings, on = 'imdb_title_id', how = 'outer',indicator = True).query("_merge == 'left_only' or _merge == 'right_only' ")
```

```python
df_movies.merge(df_ratings, on = 'imdb_title_id', how = 'outer',indicator = True)
```

left join

```python
df1.merge(df2, on = 'id',how = 'left')
```

```python
df_movies_sample = df_movies.sample(frac = 0.50)
```

```python
df_left = df_movies_sample.merge(df_ratings, on = 'imdb_title_id', how = 'left')
```

```python
df_movies_sample.shape
```

```python
df_ratings.shape
```

```python
df_left.shape
```

exclusive left join

```python
df1.merge(df2, on = 'id',how = 'outer',indicator = True).query("_merge == 'left_only'")
```

```python
df_movies2 = df_movies.copy()
```

```python
for index in df_movies2.index:
    if index <1000:
        df_movies2.loc[index,'imdb_title_id'] = 'tt1234567890'
```

```python
df_exclusive_left = df_movies2.merge(df_ratings, on = 'imdb_title_id', how = 'outer',indicator = True).query("_merge == 'left_only'")
```

```python
df_movies2.shape
```

```python
df_ratings.shape
```

```python
df_movies.shape
```

```python
df_exclusive_left.shape
```
