```python
import pandas as pd
```

```python
df_laptops = pd.read_csv('laptop_price.csv',encoding = 'unicode_escape')
```

```python
df_laptops.head(2)
```

```python
import numpy as np
```

```python
#create an array based on multiple tiers
#first creating based on conditions and value variables
conditions = [
    df_laptops['Price_euros']>3000,
    ((df_laptops['Price_euros']<=3000) &  (df_laptops['Price_euros']>2000)),
    ((df_laptops['Price_euros']<=2000) &  (df_laptops['Price_euros']>800)),
    df_laptops['Price_euros']<=800
]
values = ['too expensive','expensive','affordable','cheap']
```

```python
np.select(conditions, values,'out of expected range')
```

```python
df_laptops['Price_tiers'] = np.select(conditions, values,'out of expected range')
```

```python
df_laptops
```

```python
df_laptops.value_counts('Price_tiers')
```
