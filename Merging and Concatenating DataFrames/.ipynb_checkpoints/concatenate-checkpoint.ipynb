vertical concate

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({'id':['A','B','C','D','E'],
                    'age':[22,23,21,24,25]})
df2 = pd.DataFrame ({'id':['A','B','C','D','E'],
                    'age':[12,13,11,14,15]})
```

```python
pd.concat([df1,df2],axis =0,ignore_index = True)
```

```python
df_movies = pd.read_csv('IMDb movies.csv',low_memory = False)
df_ratings = pd.read_csv('IMDb ratings.csv')
```

```python
df_movies = df_movies[['imdb_title_id', 'title','year','genre','country']]
df_ratings = df_ratings[['imdb_title_id','total_votes', 'mean_vote']]
```

```python
df_sample = df_movies.sample(frac = 0.5)
```

```python
df_sample.shape
```

```python
df_concat_vertically = pd.concat([df_movies,df_sample],axis =0)
```

```python
df_concat_vertically.shape
```

horizontal concate

```python
df3 = pd.DataFrame({'id':['A','B','C','D','E'],
                    'age':[22,23,21,24,25]})
df4 = pd.DataFrame({'job':['Artist','Business analyst','Coder','Developer','Electrician']})
```

```python
pd.concat([df3,df4],axis = 1)
```

```python
df_movies.shape
```

```python
df_ratings.shape
```

```python
df_movies.set_index('imdb_title_id',inplace =True)
df_ratings.set_index('imdb_title_id',inplace =True)
```

```python
df_concat_horizontal = pd.concat([df_movies,df_ratings],axis = 1)
```

```python
df_concat_horizontal 
```
