```python
import pandas as pd
```

```python
df_exams = pd.read_csv('studentperformance.csv')
```

```python
df_exams.head()
```

sort a dataframe

```python
#sort one column
df_exams.sort_values(by = 'Math')
```

```python
df_exams.sort_values('Math')
```

```python
#sort descending by one column
df_exams.sort_values('Math', ascending = False)
```

```python
#sort descending by multiple columns
df_exams.sort_values(['Math','Science'], ascending = False)
```

```python
#sorting and updating dataframe using inplace --> or we can also use rewriting or reassigning to same dataframe
df_exams.sort_values(['Math','Science'], ascending = False, inplace = True)
```

```python
df_exams
```

```python
#sort descending using key function
df_exams.sort_values('Name',ascending = True, key = lambda col:col.str.lower())
```
