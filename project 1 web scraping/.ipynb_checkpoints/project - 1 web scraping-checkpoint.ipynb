Reading .csv from multiple url's

for multiple leagues

```python
#link structure
"https://football-data.co.uk/mmz4281/"+"2627"+"/"+"E0"+".csv"
```

```python
#creating root variable
root = "https://football-data.co.uk/mmz4281/"
```

```python
#creating list of leagues
leagues = ['E0','E1','E2','E3','EC']
frames =[]
#looping
for league in leagues:
    df = pd.read_csv(root + '2627' + '/' + league +'.csv', encoding = 'unicode_escape')
    frames.append(df)
```

```python
#length of frame
len(frames)
```

```python
#show elements
frames[4]
```

for multiple seasons

```python
#through the range of seasons and insert new column
#creating list of leagues
leagues = ['E0', 'E2','E3']
frames =[]
#looping
for league in leagues:
    for season in range(15,27):
        df = pd.read_csv(root + str(season) + str(season+1) + '/' + league +'.csv',encoding = 'unicode_escape' )
        df.insert(1,'Season',season)
        frames.append(df)
```

```python
#length of frames
len(frames)
```

```python
#showing element
frames[35]
```

organize dictionary data

```python
dict_countries = {'Spanish La Liga':'SP1','Spanish Segunda Divison':'SP2','German Bundesliga':'D1',
                  'English Premier League':'E0','English League 1':'E1','English League 2':'E2'}
```

```python
dict_historical_data = {}

for league in dict_countries:
    frames =[]
    for season in range(15,27):
        df = pd.read_csv(root + str(season) + str(season+1) + '/' + dict_countries[league] +'.csv',encoding = 'unicode_escape' )
        df.insert(1,'Season',season)
        frames.append(df)
    df_concat = pd.concat(frames)
    dict_historical_data[league] = df_concat

```
