```python
import pandas as pd
```

```python
df_population_raw = pd.read_csv('population_total.csv')
```

```python
df_population_raw
```

reshaping

```python
#dropping null values
df_population_raw.dropna(inplace=True)
```

```python
df_pivot = df_population_raw.pivot(index='year',columns='country',values='population')
```

```python
df_pivot = df_pivot[['United States','India','China','Indonesia','Brazil']]
```

```python
df_pivot 
```

line plot

```python
df_pivot.plot(kind= 'line',xlabel='Year',ylabel='Population',title = 'Population (1955-2020)',figsize=(15,4))
```

Bar plot

```python
#one year (2020)

df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]
```

```python

df_pivot_2020
```

```python

df_pivot_2020 = df_pivot_2020.T
```

```python
df_pivot_2020.plot(kind='bar',color='orange',xlabel='Country',ylabel='population',title='Population in 2020')
```

bar plot by n variables

```python
df_pivot
```

```python
df_pivot_sample = df_pivot[df_pivot.index.isin([1980,1990,2000,2010,2020])]
```

```python
df_pivot_sample
```

```python
df_pivot_sample.plot(kind='bar',xlabel='Year',ylabel='Population',title = 'population in 1980,1990,2000,2010,2020',figsize=(8,10))
```

```python
df_pivot_2020 
```

```python
df_pivot_2020.rename(columns={2020 : '2020'},inplace = True)
```

```python
df_pivot_2020['population in %']= (df_pivot_2020['2020']/df_pivot_2020['2020'].sum()) *100
```

```python
df_pivot_2020
```

```python
df_pivot_2020.plot(kind='pie',y = 'population in %',title = 'population 2020 in %')
```
