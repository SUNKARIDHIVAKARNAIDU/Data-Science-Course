linear regression using statsmodel

```python
import statsmodels.api as sm
```

```python
#define independent and dependent variables

x = df_boston['Rooms'] #independent 
y = df_boston['Value']  #dependent
```

```python
df_boston.plot(kind = 'scatter', x = 'Rooms' , y= 'Value')
```

```python
# making regression fitting model and adding constant
x = sm.add_constant(x)
lm = sm.OLS(y,x).fit()
```

```python
lm.predict(x)
```

```python
#regression table
lm.summary()
```

```python
#linear regression equation
#a = 9.1021#(rooms coeff)
#b = -34.6706#(const coeff)
y_pred  = 9.1021 * x['Rooms'] - 34.6706
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# plotting the data points
plt.figure(figsize=(6, 4), tight_layout=True)
sns.scatterplot(x = x['Rooms'], y=y)

# plotting the line
sns.lineplot(x=x['Rooms'], y=y_pred, color='red')

# axes
plt.xlim(0)
plt.ylim(0)
plt.savefig('linear_regression')
plt.show()
```
