```python
import pandas as pd
```

```python
df_boston = pd.read_csv('Boston House Prices.csv')
df_boston
```

```python
from sklearn import linear_model
```

```python
y = df_boston['Value']
x = df_boston[['Rooms','Distance']]
```

```python
lm = linear_model.LinearRegression()
lm.fit(x,y)
```

```python
lm.predict(x)
```

```python
lm.score(x,y)
```

```python
lm.coef_
```

```python
lm.intercept_
```
